```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
from sklearn.neighbors import NearestNeighbors

import pickle, os
os.makedirs("data", exist_ok=True)
print("Libraries loaded. data/ folder ready.")
```

## Step 1 — Load Preprocessed Data

```python
#LOAD DATA
full_df      = pd.read_csv("data/cleaned_data.csv")

feature_cols = [c for c in full_df.columns if c != 'disease_label']
X_scaled     = full_df[feature_cols].values
y            = full_df['disease_label'].values

# Load label encoders to decode integer labels back to disease names
with open('data/label_encoders.pkl', 'rb') as f:
    label_encoders = pickle.load(f)
class_names = label_encoders['disease_label'].classes_

print(f"Data loaded  :  {X_scaled.shape[0]} samples  x  {X_scaled.shape[1]} features")
print(f"Disease classes : {len(class_names)}")
print(f"Classes : {list(class_names)}")

# PCA 2D for all visualisations
pca      = PCA(n_components=2, random_state=42)
X_2d     = pca.fit_transform(X_scaled)
explained = pca.explained_variance_ratio_.sum() * 100
print(f"\nPCA 2D variance explained: {explained:.1f}%")
print("(Lower % is expected with 13 features — PCA is only used for visualisation)")
```

## Step 2 — K-Means Clustering

```python
# K-MEANS CLUSTERING
print("K-Means: testing K = 2 to 12...\n")

inertias, sil_scores = [], []
K_range = range(2, 13)
for k in K_range:
    km  = KMeans(n_clusters=k, random_state=42, n_init=10)
    lbl = km.fit_predict(X_scaled)
    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(X_scaled, lbl))

BEST_K = list(K_range)[np.argmax(sil_scores)]
print(f"Best K (highest silhouette): {BEST_K}  (score = {max(sil_scores):.4f})")

# Fit final K-Means model with best K
kmeans    = KMeans(n_clusters=BEST_K, random_state=42, n_init=10)
km_labels = kmeans.fit_predict(X_scaled)
sil_km    = silhouette_score(X_scaled, km_labels)

print(f"\nCluster sizes:")
for k, cnt in pd.Series(km_labels).value_counts().sort_index().items():
    # Decode the most common disease in each cluster for interpretability
    top_disease_idx = pd.Series(y[km_labels == k]).mode()[0]
    top_disease     = label_encoders['disease_label'].inverse_transform([top_disease_idx])[0]
    print(f"  Cluster {k} : {cnt:3d} samples  (dominant disease: {top_disease})")
```

```python
# K-Means charts: elbow, silhouette, PCA scatter, disease proportion per cluster
colors_cl = ['#E74C3C', '#3498DB', '#2ECC71', '#9B59B6', '#F39C12',
             '#1ABC9C', '#E67E22', '#8E44AD', '#2C3E50', '#16A085',
             '#D35400', '#C0392B', '#27AE60']

fig, axes = plt.subplots(2, 2, figsize=(13, 10))
fig.suptitle("K-Means Clustering Results", fontsize=14, fontweight='bold')

# Elbow plot 
axes[0, 0].plot(K_range, inertias, 'bo-', markersize=7, linewidth=2)
axes[0, 0].axvline(BEST_K, color='red', linestyle='--', label=f'Best K={BEST_K}')
axes[0, 0].set_title("Elbow Method (Inertia)")
axes[0, 0].legend()
axes[0, 0].set_xlabel("K"); axes[0, 0].set_ylabel("Inertia (WCSS)")

# Silhouette score per K
axes[0, 1].plot(K_range, sil_scores, 'gs-', markersize=7, linewidth=2)
axes[0, 1].axvline(BEST_K, color='red', linestyle='--', label=f'Best K={BEST_K}')
axes[0, 1].set_title("Silhouette Score per K")
axes[0, 1].legend()
axes[0, 1].set_xlabel("K"); axes[0, 1].set_ylabel("Silhouette Score")

# PCA 2D scatter 
for cl in range(BEST_K):
    mask = km_labels == cl
    axes[1, 0].scatter(X_2d[mask, 0], X_2d[mask, 1],
                       c=colors_cl[cl], label=f"Cluster {cl} (n={mask.sum()})",
                       alpha=0.6, s=30, edgecolors='none')
axes[1, 0].set_title(f"K-Means Clusters in PCA 2D  ({explained:.1f}% variance)")
axes[1, 0].set_xlabel("PC1"); axes[1, 0].set_ylabel("PC2")
axes[1, 0].legend(fontsize=8)

# Dominant disease per cluster 
full_df_raw = pd.read_csv("data/cleaned_data.csv")
full_df_raw['KMeans_Cluster'] = km_labels

dominant_pct = []
dominant_names = []
for cl in range(BEST_K):
    subset = full_df_raw[full_df_raw['KMeans_Cluster'] == cl]['disease_label']
    top_idx  = subset.mode()[0]
    top_name = label_encoders['disease_label'].inverse_transform([top_idx])[0]
    top_pct  = (subset == top_idx).mean() * 100
    dominant_pct.append(top_pct)
    dominant_names.append(top_name)

bars = axes[1, 1].bar(range(BEST_K), dominant_pct,
                      color=colors_cl[:BEST_K], edgecolor='white', width=0.5)
for bar, val, name in zip(bars, dominant_pct, dominant_names):
    axes[1, 1].text(bar.get_x() + bar.get_width()/2, val + 0.5,
                    f"{val:.1f}%\n({name})", ha='center',
                    fontweight='bold', fontsize=8)
axes[1, 1].set_title("Dominant Disease % per K-Means Cluster")
axes[1, 1].set_xlabel("Cluster"); axes[1, 1].set_ylabel("Dominant Disease (%)")
axes[1, 1].set_ylim(0, 120)
axes[1, 1].set_xticks(range(BEST_K))

plt.tight_layout()
plt.savefig("data/kmeans_results.png", dpi=130, bbox_inches='tight')
plt.show()
print("Saved: data/kmeans_results.png")
```

## Step 3 — DBSCAN Clustering

```python
# STEP 3 : DBSCAN CLUSTERING
nbrs = NearestNeighbors(n_neighbors=5).fit(X_scaled)
distances, _ = nbrs.kneighbors(X_scaled)
k_dist_sorted = np.sort(distances[:, 4])

fig_eps, ax_eps = plt.subplots(figsize=(8, 4))
ax_eps.plot(k_dist_sorted, color='steelblue', linewidth=2)
ax_eps.axhline(2.3, color='red', linestyle='--', linewidth=1.8,
               label='Chosen eps = 2.3')
ax_eps.set_title("K-Distance Graph (k=5) — eps Justification for DBSCAN",
                 fontweight='bold')
ax_eps.set_xlabel("Points sorted by 5-NN distance (ascending)")
ax_eps.set_ylabel("5-NN Distance")
ax_eps.legend()
plt.tight_layout()
plt.savefig("data/dbscan_eps.png", dpi=130, bbox_inches='tight')
plt.show()
print("Saved: data/dbscan_eps.png  (elbow at ~2.3 justifies eps=2.3)")

# Fit DBSCAN
dbscan    = DBSCAN(eps=2.3, min_samples=5)
db_labels = dbscan.fit_predict(X_scaled)
n_db_clusters = len(set(db_labels)) - (1 if -1 in db_labels else 0)
n_noise       = list(db_labels).count(-1)

print(f"\nDBSCAN  |  eps=2.3  |  min_samples=5")
print(f"  Clusters found : {n_db_clusters}")
print(f"  Noise points   : {n_noise}  ({n_noise/len(db_labels)*100:.1f}%)")
print(f"\n  Cluster breakdown:")
for lbl in sorted(set(db_labels)):
    cnt = (db_labels == lbl).sum()
    if lbl == -1:
        print(f"    Noise           : {cnt:3d} samples")
    else:
        subset      = y[db_labels == lbl]
        top_idx     = pd.Series(subset).mode()[0]
        top_disease = label_encoders['disease_label'].inverse_transform([top_idx])[0]
        print(f"    Cluster {lbl}       : {cnt:3d} samples  (dominant: {top_disease})")

if n_db_clusters > 1:
    valid  = db_labels != -1
    sil_db = silhouette_score(X_scaled[valid], db_labels[valid])
    print(f"\n  Silhouette (non-noise) : {sil_db:.4f}")
```

```python
# DBSCAN visualisation
fig4, axes4 = plt.subplots(1, 2, figsize=(13, 5))
fig4.suptitle("DBSCAN Clustering Results", fontsize=14, fontweight='bold')

unique_labels = sorted(set(db_labels))
tab_colors    = plt.cm.tab10(np.linspace(0, 1, max(len(unique_labels), 3)))

for i, lbl in enumerate(unique_labels):
    mask  = db_labels == lbl
    label = f"Noise ({mask.sum()})" if lbl == -1 else f"Cluster {lbl} ({mask.sum()})"
    clr   = 'grey' if lbl == -1 else tab_colors[i]
    alpha = 0.25   if lbl == -1 else 0.7
    axes4[0].scatter(X_2d[mask, 0], X_2d[mask, 1],
                     c=[clr], label=label, alpha=alpha, s=25, edgecolors='none')
axes4[0].set_title(f"DBSCAN — PCA 2D  ({n_db_clusters} clusters | {n_noise} noise)")
axes4[0].set_xlabel("PC1"); axes4[0].set_ylabel("PC2")
axes4[0].legend(fontsize=8)

lbl_counts = pd.Series(db_labels).value_counts().sort_index()
bar_clrs   = ['grey' if l == -1 else 'steelblue' for l in lbl_counts.index]
axes4[1].bar([str(l) if l != -1 else 'Noise' for l in lbl_counts.index],
             lbl_counts.values, color=bar_clrs, edgecolor='white')
axes4[1].set_title("Points per DBSCAN Label")
axes4[1].set_xlabel("Cluster Label  (-1 = Noise)"); axes4[1].set_ylabel("Count")

plt.tight_layout()
plt.savefig("data/dbscan_results.png", dpi=130, bbox_inches='tight')
plt.show()
print("Saved: data/dbscan_results.png")
```

## Step 4 — Clustering Comparison & Save

```python
# STEP 4 : CLUSTERING COMPARISON 
sil_km = silhouette_score(X_scaled, km_labels)
if n_db_clusters > 1:
    valid      = db_labels != -1
    sil_db     = silhouette_score(X_scaled[valid], db_labels[valid])
    sil_db_str = f"{sil_db:.4f}"
else:
    sil_db_str = "N/A (1 cluster)"

print("\nClustering Algorithm Comparison:")
print(f"{'Algorithm':<30} {'Silhouette':>12} {'Clusters':>10} {'Noise':>8}")
print("-" * 65)
print(f"{'K-Means (k='+str(BEST_K)+')':<30} {sil_km:>12.4f} {BEST_K:>10} {'N/A':>8}")
print(f"{'DBSCAN (eps=2.3, ms=5)':<30} {sil_db_str:>12} {n_db_clusters:>10} {n_noise:>8}")
print(f"\nInterpretation:")
print(f"  Silhouette score ranges from -1 to +1.")
print(f"  Higher = better-separated, more cohesive clusters.")
print(f"  K-Means forces every point into a cluster (no noise label).")
print(f"  DBSCAN identifies {n_noise} outlier crop samples as noise —")
print(f"  these may represent rare disease conditions or measurement errors.")

# Side-by-side PCA comparison
fig_cmp, axes_cmp = plt.subplots(1, 2, figsize=(14, 5))
fig_cmp.suptitle("Clustering Comparison — K-Means vs DBSCAN (PCA 2D)",
                 fontsize=13, fontweight='bold')

# K-Means panel
colors_cl = ['#E74C3C', '#3498DB', '#2ECC71', '#9B59B6', '#F39C12',
             '#1ABC9C', '#E67E22', '#8E44AD', '#2C3E50', '#16A085',
             '#D35400', '#C0392B', '#27AE60']
for cl in range(BEST_K):
    mask = km_labels == cl
    axes_cmp[0].scatter(X_2d[mask, 0], X_2d[mask, 1],
                        c=colors_cl[cl], label=f"Cluster {cl}",
                        alpha=0.6, s=25)
axes_cmp[0].set_title(f"K-Means  (K={BEST_K}, Silhouette={sil_km:.4f})")
axes_cmp[0].set_xlabel("PC1"); axes_cmp[0].set_ylabel("PC2")
axes_cmp[0].legend(fontsize=8)

# DBSCAN panel
for i, lbl in enumerate(sorted(set(db_labels))):
    mask  = db_labels == lbl
    label = "Noise" if lbl == -1 else f"Cluster {lbl}"
    clr   = 'grey' if lbl == -1 else tab_colors[i]
    alpha = 0.2    if lbl == -1 else 0.7
    axes_cmp[1].scatter(X_2d[mask, 0], X_2d[mask, 1],
                        c=[clr], label=label, alpha=alpha, s=25)
axes_cmp[1].set_title(f"DBSCAN  ({n_db_clusters} clusters, {n_noise} noise, Sil={sil_db_str})")
axes_cmp[1].set_xlabel("PC1"); axes_cmp[1].set_ylabel("PC2")
axes_cmp[1].legend(fontsize=8)

plt.tight_layout()
plt.savefig("data/clustering_comparison.png", dpi=130, bbox_inches='tight')
plt.show()
print("Saved: data/clustering_comparison.png")

# Save cluster assignments back to CSV
result_df = pd.read_csv("data/cleaned_data.csv")
result_df['KMeans_Cluster'] = km_labels
result_df['DBSCAN_Cluster'] = db_labels
result_df.to_csv("data/cluster_results.csv", index=False)
print("Saved: data/cluster_results.csv")
print("\n✓ Clustering complete.")
```
