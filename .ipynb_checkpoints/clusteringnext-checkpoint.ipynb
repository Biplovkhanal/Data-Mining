```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
from sklearn.neighbors import NearestNeighbors

import pickle, os
os.makedirs("data", exist_ok=True)
print("Libraries loaded. data/ folder ready.")
```

## Step 1 — Load Preprocessed Data

```
# ─────────────────────────────────────────────────────────────
# STEP 1 : LOAD PREPROCESSED DATA
# ─────────────────────────────────────────────────────────────
# ⚠️ IMPORTANT — NO preprocessing is done here.
#   All cleaning, imputation, outlier capping, encoding, and scaling
#   was already completed in data_preparation.ipynb and saved to data/.
#   This notebook simply loads those ready-made files.
#
# Files used:
#   data/cleaned_data.csv      — full scaled dataset (all 13 features + label)
#   data/label_encoders.pkl    — LabelEncoder objects to decode integers back
#                                to original text (e.g. 0 → 'Hilly')
#   data/scaler.pkl            — StandardScaler + feature column order
#
# Goal: Cluster farming regions based on AGRICULTURAL DECISION features only.
#   Features used  : region, crop_type, fertilizer_used, soil_ph, nitrogen,
#                    phosphorus, potassium, temperature, humidity, rainfall
#   Features excluded : disease_label, disease_severity, npk_ratio,
#                       heat_humidity_index
#   Reason: disease outcomes are what we PREDICT in classification.ipynb.
#   Clustering answers a different question:
#   "Which farms share similar agricultural conditions so we can give
#    zone-level recommendations on irrigation, fertilizer, and crop choice?"
# ─────────────────────────────────────────────────────────────

# Load the full cleaned & scaled dataset
cleaned_df = pd.read_csv("data/cleaned_data.csv")

# Load label encoders so we can decode columns back to readable text
with open('data/label_encoders.pkl', 'rb') as f:
    label_encoders = pickle.load(f)

# Load scaler to know the exact column order used during preprocessing
with open('data/scaler.pkl', 'rb') as f:
    scaler, all_feature_cols = pickle.load(f)

print(f"Cleaned dataset shape : {cleaned_df.shape}")
print(f"All scaled columns    : {list(cleaned_df.columns)}")

# ── Select ONLY agricultural decision features ────────────────
# These 10 columns represent farming environment & decisions.
# disease_label, disease_severity, npk_ratio, heat_humidity_index are excluded.
AGRI_FEATURES = [
    'region_enc', 'crop_enc', 'fert_enc',
    'soil_ph', 'nitrogen', 'phosphorus', 'potassium',
    'temperature', 'humidity', 'rainfall'
]

X_agri   = cleaned_df[AGRI_FEATURES].values   # shape: (548, 10) — already scaled
y_disease = cleaned_df['disease_label'].values  # kept ONLY for post-hoc check

print(f"\nAgricultural feature matrix : {X_agri.shape}")
print(f"Features used : {AGRI_FEATURES}")

# ── Decode encodings for readable labels ─────────────────────
# The preprocessing encoded categories as integers (e.g. region 0,1,2).
# We reverse this here so plots show real names (Terai, Hilly, Mountain).
le_region = label_encoders['region']
le_crop   = label_encoders['crop_type']
le_fert   = label_encoders['fertilizer_used']

region_col = le_region.inverse_transform(cleaned_df['region_enc'].astype(int).values)
crop_col   = le_crop.inverse_transform(cleaned_df['crop_enc'].astype(int).values)
fert_col   = le_fert.inverse_transform(cleaned_df['fert_enc'].astype(int).values)

# Build a decoded dataframe for readable EDA and post-hoc plots
df = cleaned_df.copy()
df['region']          = region_col
df['crop_type']       = crop_col
df['fertilizer_used'] = fert_col
df['disease_label']   = label_encoders['disease_label'].inverse_transform(
                            cleaned_df['disease_label'].astype(int).values)

print(f"\nSamples per region:")
for region, cnt in pd.Series(region_col).value_counts().items():
    print(f"  {region:10s} : {cnt} samples")

print(f"\nSamples per crop type:")
for crop, cnt in pd.Series(crop_col).value_counts().items():
    print(f"  {crop:10s} : {cnt} samples")

# ── EDA — feature distributions by region ────────────────────
# These are the ALREADY SCALED values from cleaned_data.csv.
# We use original-scale columns (temperature, humidity, etc.) decoded above.
env_cols     = ['temperature', 'humidity', 'rainfall', 'soil_ph']
env_names    = ['Temperature (°C)', 'Humidity (%)', 'Rainfall (mm)', 'Soil pH']
region_palette = {'Terai': '#E74C3C', 'Hilly': '#3498DB', 'Mountain': '#2ECC71'}

fig, axes = plt.subplots(2, 2, figsize=(13, 9))
fig.suptitle("Agricultural Feature Distributions by Farming Region",
             fontsize=14, fontweight='bold')
axes = axes.flatten()

for i, (col, name) in enumerate(zip(env_cols, env_names)):
    for region in ['Terai', 'Hilly', 'Mountain']:
        mask = df['region'] == region
        axes[i].hist(cleaned_df.loc[mask, col], bins=20, alpha=0.55,
                     label=region, color=region_palette[region], edgecolor='white')
    axes[i].set_title(name)
    axes[i].set_xlabel('Scaled Value (from data_preparation.ipynb)')
    axes[i].set_ylabel('Frequency')
    axes[i].legend(fontsize=8)

plt.tight_layout()
plt.savefig("data/eda_agri_region.png", dpi=130, bbox_inches='tight')
plt.show()
print("  Saved: data/eda_agri_region.png")

# ── PCA 2D for all visualisations ────────────────────────────
# ⚠️ PCA is applied ONLY for 2D scatter plots — the actual clustering
#    runs on the full 10-dimensional X_agri matrix.
pca      = PCA(n_components=2, random_state=42)
X_2d     = pca.fit_transform(X_agri)
explained = pca.explained_variance_ratio_.sum() * 100
print(f"\nPCA 2D variance explained: {explained:.1f}%")
print("  (Used only for scatter plots; clustering runs on full 10D feature space)")
```

## Step 2 — K-Means Clustering

```
# ─────────────────────────────────────────────────────────────
# STEP 2 : K-MEANS CLUSTERING
# ─────────────────────────────────────────────────────────────
# K-Means partitions all 548 farms into K agricultural zones by
# minimising Within-Cluster Sum of Squares (WCSS / Inertia) across
# all 10 farming features.
# We sweep K = 2..12 and pick the K with the highest Silhouette Score.
#
# ⚠️ CONCEPT — Silhouette Score:
#   Measures how similar a sample is to its own cluster vs. other clusters.
#   Range: -1 (wrong cluster) to +1 (perfect separation). Higher = better.
print("K-Means: testing K = 2 to 12...\n")

inertias, sil_scores = [], []
K_range = range(2, 13)

for k in K_range:
    km  = KMeans(n_clusters=k, random_state=42, n_init=10)
    lbl = km.fit_predict(X_agri)
    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(X_agri, lbl))
    print(f"  K={k:2d}  |  Inertia={km.inertia_:8.1f}  |  Silhouette={sil_scores[-1]:.4f}")

BEST_K = list(K_range)[np.argmax(sil_scores)]
print(f"\nBest K (highest Silhouette) : {BEST_K}  (score = {max(sil_scores):.4f})")

# Fit final model with best K
kmeans    = KMeans(n_clusters=BEST_K, random_state=42, n_init=10)
km_labels = kmeans.fit_predict(X_agri)
sil_km    = silhouette_score(X_agri, km_labels)

df['KMeans_Zone'] = km_labels

print(f"\nFarming Zone Summary:")
print(f"{'Zone':<8} {'Samples':>8} {'Dominant Region':>18} {'Dominant Crop':>16} {'Avg Temp (C)':>14} {'Avg Rain (mm)':>14}")
print("-" * 82)
for k in range(BEST_K):
    subset     = df[df['KMeans_Zone'] == k]
    dom_region = subset['region'].mode()[0]
    dom_crop   = subset['crop_type'].mode()[0]
    avg_temp   = subset['temperature'].mean()
    avg_rain   = subset['rainfall'].mean()
    print(f"  {k:<6} {len(subset):>8} {dom_region:>18} {dom_crop:>16} {avg_temp:>14.3f} {avg_rain:>14.3f}")
print("  (temperature and rainfall are in scaled units from StandardScaler)")
```

```
# K-Means charts: elbow, silhouette, PCA scatter, fertilizer rate per zone
ZONE_COLORS = ['#E74C3C', '#3498DB', '#2ECC71', '#9B59B6', '#F39C12',
               '#1ABC9C', '#E67E22', '#8E44AD', '#2C3E50', '#16A085',
               '#D35400', '#C0392B', '#27AE60']

fig, axes = plt.subplots(2, 2, figsize=(13, 10))
fig.suptitle("K-Means Farming Zone Clustering Results", fontsize=14, fontweight='bold')

# ── Elbow ─────────────────────────────────────────────────────
axes[0, 0].plot(K_range, inertias, 'bo-', markersize=7, linewidth=2)
axes[0, 0].axvline(BEST_K, color='red', linestyle='--', label=f'Best K={BEST_K}')
axes[0, 0].set_title("Elbow Method (Inertia)")
axes[0, 0].set_xlabel("K"); axes[0, 0].set_ylabel("Inertia (WCSS)")
axes[0, 0].legend()

# ── Silhouette ────────────────────────────────────────────────
axes[0, 1].plot(K_range, sil_scores, 'gs-', markersize=7, linewidth=2)
axes[0, 1].axvline(BEST_K, color='red', linestyle='--', label=f'Best K={BEST_K}')
axes[0, 1].set_title("Silhouette Score per K")
axes[0, 1].set_xlabel("K"); axes[0, 1].set_ylabel("Silhouette Score")
axes[0, 1].legend()

# ── PCA scatter coloured by farming zone ──────────────────────
for zone in range(BEST_K):
    mask     = km_labels == zone
    dom_crop = df[df['KMeans_Zone'] == zone]['crop_type'].mode()[0]
    axes[1, 0].scatter(X_2d[mask, 0], X_2d[mask, 1],
                       c=ZONE_COLORS[zone],
                       label=f"Zone {zone} — {dom_crop} ({mask.sum()})",
                       alpha=0.65, s=35, edgecolors='none')
axes[1, 0].set_title(f"Farming Zones in PCA 2D  ({explained:.1f}% variance captured)")
axes[1, 0].set_xlabel("PC1 (Climate / Rainfall axis)")
axes[1, 0].set_ylabel("PC2 (Soil / Crop axis)")
axes[1, 0].legend(fontsize=8)

# ── Fertilizer usage rate per zone ────────────────────────────
# Agricultural insight: how many farms in each zone use fertilizer?
# This directly drives zone-specific fertilizer policy recommendations.
fert_rates = []
for zone in range(BEST_K):
    subset    = df[df['KMeans_Zone'] == zone]
    fert_rate = (subset['fertilizer_used'] == 'Yes').mean() * 100
    fert_rates.append(fert_rate)

bars = axes[1, 1].bar(range(BEST_K), fert_rates,
                      color=ZONE_COLORS[:BEST_K], edgecolor='white', width=0.5)
for bar, val, zone in zip(bars, fert_rates, range(BEST_K)):
    dom_crop = df[df['KMeans_Zone'] == zone]['crop_type'].mode()[0]
    axes[1, 1].text(bar.get_x() + bar.get_width()/2, val + 0.5,
                    f"{val:.1f}%\n({dom_crop})",
                    ha='center', fontweight='bold', fontsize=9)
axes[1, 1].set_title("Fertilizer Usage Rate (%) per Farming Zone")
axes[1, 1].set_xlabel("Farming Zone")
axes[1, 1].set_ylabel("Farms Using Fertilizer (%)")
axes[1, 1].set_ylim(0, 120)
axes[1, 1].set_xticks(range(BEST_K))

plt.tight_layout()
plt.savefig("data/kmeans_agri_results.png", dpi=130, bbox_inches='tight')
plt.show()
print("Saved: data/kmeans_agri_results.png")
```

```
# ── K-Means Zone Profile Deep-Dive ───────────────────────────
fig2, axes2 = plt.subplots(1, 2, figsize=(16, 5))
fig2.suptitle("K-Means Farming Zone Profiles", fontsize=13, fontweight='bold')

# Heatmap of mean scaled feature values per zone
num_feats    = ['soil_ph', 'nitrogen', 'phosphorus', 'potassium',
                'temperature', 'humidity', 'rainfall']
zone_profile = cleaned_df.copy()
zone_profile['KMeans_Zone'] = km_labels
zone_profile = zone_profile.groupby('KMeans_Zone')[num_feats].mean().round(3)
sns.heatmap(zone_profile, annot=True, fmt='.2f', cmap='YlOrRd',
            linewidths=0.4, ax=axes2[0], annot_kws={"size": 9})
axes2[0].set_title("Mean Scaled Feature Values per Farming Zone")
axes2[0].set_ylabel("Farming Zone")
axes2[0].set_xlabel("(values are StandardScaler-scaled from data_preparation.ipynb)")

# Stacked bar: crop composition per zone
crop_zone     = df.groupby(['KMeans_Zone', 'crop_type']).size().unstack(fill_value=0)
crop_zone_pct = crop_zone.div(crop_zone.sum(axis=1), axis=0) * 100
crop_palette  = sns.color_palette("tab10", len(crop_zone_pct.columns))
bottom        = np.zeros(BEST_K)
for j, crop in enumerate(crop_zone_pct.columns):
    vals = crop_zone_pct[crop].values
    axes2[1].bar(range(BEST_K), vals, bottom=bottom,
                 color=crop_palette[j], label=crop, edgecolor='white', width=0.55)
    for k_idx, (v, b) in enumerate(zip(vals, bottom)):
        if v > 5:
            axes2[1].text(k_idx, b + v/2, f"{v:.0f}%",
                          ha='center', va='center', fontsize=7.5,
                          color='white', fontweight='bold')
    bottom += vals
axes2[1].set_title("Crop Composition per Farming Zone (%)")
axes2[1].set_xlabel("Farming Zone"); axes2[1].set_ylabel("Percentage (%)")
axes2[1].set_xticks(range(BEST_K))
axes2[1].legend(fontsize=8, loc='upper right', bbox_to_anchor=(1.18, 1))

plt.tight_layout()
plt.savefig("data/kmeans_zone_profiles.png", dpi=130, bbox_inches='tight')
plt.show()
print("Saved: data/kmeans_zone_profiles.png")

# Post-hoc: disease check per zone (disease_label was NOT used in clustering)
print("\n[Post-hoc] Top diseases per farming zone (NOT used in clustering):")
print("-" * 65)
for zone in range(BEST_K):
    subset  = df[df['KMeans_Zone'] == zone]
    dom_reg = subset['region'].mode()[0]
    top_dis = subset['disease_label'].value_counts().head(3).to_dict()
    print(f"  Zone {zone} | region: {dom_reg} | top diseases: {top_dis}")
```

## Step 3 — DBSCAN Clustering

```
# ─────────────────────────────────────────────────────────────
# STEP 3 : DBSCAN CLUSTERING
# ─────────────────────────────────────────────────────────────
# DBSCAN (Density-Based Spatial Clustering of Applications with Noise)
# discovers farming micro-zones of arbitrary shape without specifying K.
# Farms in sparse regions are labelled -1 (NOISE) — these represent
# ATYPICAL farming conditions that do not fit any mainstream zone.
#
# ⚠️ CONCEPT — Key hyperparameters:
#   eps         — neighbourhood radius: two farms are "neighbours" if
#                 their scaled-feature distance <= eps
#   min_samples — minimum neighbours required to form a dense core point

# ── 3.1  K-Distance Graph to choose eps ──────────────────────
# Sort samples by their 5th-nearest-neighbour distance (ascending).
# The 'elbow' (sharp bend) separates dense core points from sparse outliers.
# We set eps at that elbow so that only genuinely sparse farms become noise.
nbrs          = NearestNeighbors(n_neighbors=5).fit(X_agri)
distances, _  = nbrs.kneighbors(X_agri)
k_dist_sorted = np.sort(distances[:, 4])

fig_eps, ax_eps = plt.subplots(figsize=(8, 4))
ax_eps.plot(k_dist_sorted, color='steelblue', linewidth=2)
ax_eps.axhline(1.8, color='red', linestyle='--', linewidth=1.8,
               label='Chosen eps = 1.8  (elbow point ~70th percentile)')
ax_eps.set_title("K-Distance Graph (k=5) — eps Selection for DBSCAN",
                 fontsize=12, fontweight='bold')
ax_eps.set_xlabel("Samples sorted by 5-NN distance (ascending)")
ax_eps.set_ylabel("5-NN Distance")
ax_eps.legend()
plt.tight_layout()
plt.savefig("data/dbscan_eps.png", dpi=130, bbox_inches='tight')
plt.show()
print("Saved: data/dbscan_eps.png")
print("  eps = 1.8 chosen at the graph elbow (70th percentile of 5-NN distances).")

# ── 3.2  Fit DBSCAN ──────────────────────────────────────────
dbscan    = DBSCAN(eps=1.8, min_samples=5)
db_labels = dbscan.fit_predict(X_agri)
df['DBSCAN_Zone'] = db_labels

n_db_clusters = len(set(db_labels)) - (1 if -1 in db_labels else 0)
n_noise       = (db_labels == -1).sum()

print(f"\nDBSCAN  |  eps=1.8  |  min_samples=5")
print(f"  Micro-zones found : {n_db_clusters}")
print(f"  Atypical farms    : {n_noise}  ({n_noise/len(db_labels)*100:.1f}%)")

if n_db_clusters > 1:
    valid  = db_labels != -1
    sil_db = silhouette_score(X_agri[valid], db_labels[valid])
    print(f"  Silhouette (valid): {sil_db:.4f}")

print(f"\n  Micro-zone breakdown:")
print(f"  {'Zone':<8} {'Samples':>8} {'Dominant Region':>18} {'Dominant Crop':>16}")
print("  " + "-" * 56)
for lbl in sorted(set(db_labels)):
    subset = df[df['DBSCAN_Zone'] == lbl]
    if lbl == -1:
        print(f"  {'Noise':<8} {len(subset):>8}   (atypical farms not fitting any zone)")
    else:
        dom_reg  = subset['region'].mode()[0]
        dom_crop = subset['crop_type'].mode()[0]
        print(f"  {lbl:<8} {len(subset):>8} {dom_reg:>18} {dom_crop:>16}")
```

```
# DBSCAN visualisation
tab_colors    = plt.cm.tab10(np.linspace(0, 1, max(n_db_clusters + 1, 3)))
unique_labels = sorted(set(db_labels))

fig4, axes4 = plt.subplots(1, 2, figsize=(14, 5))
fig4.suptitle("DBSCAN Agricultural Micro-Zone Results", fontsize=14, fontweight='bold')

# PCA scatter
for i, lbl in enumerate(unique_labels):
    mask = db_labels == lbl
    if lbl == -1:
        label  = f"Atypical Farms — Noise ({mask.sum()})"
        clr, alpha, marker = 'grey', 0.25, 'x'
    else:
        dom_crop = df[df['DBSCAN_Zone'] == lbl]['crop_type'].mode()[0]
        label    = f"Zone {lbl} — {dom_crop} ({mask.sum()})"
        clr, alpha, marker = tab_colors[i], 0.75, 'o'
    axes4[0].scatter(X_2d[mask, 0], X_2d[mask, 1],
                     c=[clr], label=label, alpha=alpha,
                     s=30, marker=marker, edgecolors='none')
axes4[0].set_title(f"DBSCAN — PCA 2D  ({n_db_clusters} micro-zones | {n_noise} atypical farms)")
axes4[0].set_xlabel("PC1 (Climate / Rainfall axis)")
axes4[0].set_ylabel("PC2 (Soil / Crop axis)")
axes4[0].legend(fontsize=7.5)

# Sample count bar per zone
lbl_counts = pd.Series(db_labels).value_counts().sort_index()
bar_labels = ['Noise' if l == -1 else f"Zone {l}" for l in lbl_counts.index]
bar_clrs   = ['grey' if l == -1 else tab_colors[i]
               for i, l in enumerate(lbl_counts.index)]
axes4[1].bar(bar_labels, lbl_counts.values, color=bar_clrs, edgecolor='white')
for j, (bar_lbl, cnt) in enumerate(zip(bar_labels, lbl_counts.values)):
    if bar_lbl != 'Noise':
        zone_id  = int(bar_lbl.split()[1])
        dom_crop = df[df['DBSCAN_Zone'] == zone_id]['crop_type'].mode()[0]
        axes4[1].text(j, cnt + 1, dom_crop, ha='center', fontsize=7.5, rotation=15)
axes4[1].set_title("Samples per DBSCAN Zone")
axes4[1].set_xlabel("Zone  (Noise = atypical farming conditions)")
axes4[1].set_ylabel("Count")
axes4[1].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.savefig("data/dbscan_agri_results.png", dpi=130, bbox_inches='tight')
plt.show()
print("Saved: data/dbscan_agri_results.png")
```

## Step 4 — Comparison & Agricultural Decision Summary

```
# ─────────────────────────────────────────────────────────────
# STEP 4 : COMPARISON & AGRICULTURAL DECISION SUMMARY
# ─────────────────────────────────────────────────────────────

sil_km = silhouette_score(X_agri, km_labels)
if n_db_clusters > 1:
    valid      = db_labels != -1
    sil_db     = silhouette_score(X_agri[valid], db_labels[valid])
    sil_db_str = f"{sil_db:.4f}"
else:
    sil_db_str = "N/A"

print("Clustering Algorithm Comparison — Agricultural Zones")
print(f"{'Algorithm':<35} {'Silhouette':>12} {'Zones':>7} {'Noise':>8}")
print("-" * 65)
print(f"{'K-Means (K='+str(BEST_K)+')':<35} {sil_km:>12.4f} {BEST_K:>7} {'None':>8}")
print(f"{'DBSCAN (eps=1.8, min_samples=5)':<35} {sil_db_str:>12} {n_db_clusters:>7} {n_noise:>8}")

print("\nInterpretation:")
print(f"  K-Means : {BEST_K} broad farming zones; every farm assigned to a zone.")
print(f"            Best for region-wide policy planning (irrigation, crop rotation).")
print(f"  DBSCAN  : {n_db_clusters} micro-zones + {n_noise} atypical farms flagged as noise.")
print(f"            Better for finding niche pockets and farms needing special attention.")

# Side-by-side comparison plot
fig_cmp, axes_cmp = plt.subplots(1, 2, figsize=(14, 5))
fig_cmp.suptitle("Farming Region Clustering Comparison — K-Means vs DBSCAN",
                 fontsize=13, fontweight='bold')

for zone in range(BEST_K):
    mask     = km_labels == zone
    dom_crop = df[df['KMeans_Zone'] == zone]['crop_type'].mode()[0]
    axes_cmp[0].scatter(X_2d[mask, 0], X_2d[mask, 1],
                        c=ZONE_COLORS[zone],
                        label=f"Zone {zone} — {dom_crop}",
                        alpha=0.65, s=28)
axes_cmp[0].set_title(f"K-Means  (K={BEST_K}, Silhouette={sil_km:.4f})")
axes_cmp[0].set_xlabel("PC1"); axes_cmp[0].set_ylabel("PC2")
axes_cmp[0].legend(fontsize=8)

for i, lbl in enumerate(sorted(set(db_labels))):
    mask = db_labels == lbl
    if lbl == -1:
        axes_cmp[1].scatter(X_2d[mask, 0], X_2d[mask, 1],
                            c='grey', label=f"Noise ({mask.sum()})",
                            alpha=0.2, s=20, marker='x')
    else:
        dom_crop = df[df['DBSCAN_Zone'] == lbl]['crop_type'].mode()[0]
        axes_cmp[1].scatter(X_2d[mask, 0], X_2d[mask, 1],
                            c=[tab_colors[i]], alpha=0.7, s=28,
                            label=f"Zone {lbl} — {dom_crop}")
axes_cmp[1].set_title(f"DBSCAN  ({n_db_clusters} micro-zones | {n_noise} noise | Sil={sil_db_str})")
axes_cmp[1].set_xlabel("PC1"); axes_cmp[1].set_ylabel("PC2")
axes_cmp[1].legend(fontsize=7.5)

plt.tight_layout()
plt.savefig("data/clustering_agri_comparison.png", dpi=130, bbox_inches='tight')
plt.show()
print("Saved: data/clustering_agri_comparison.png")

# Save cluster assignments
result_df = df[['crop_type', 'region', 'fertilizer_used', 'soil_ph',
                'nitrogen', 'phosphorus', 'potassium',
                'temperature', 'humidity', 'rainfall',
                'disease_label', 'KMeans_Zone', 'DBSCAN_Zone']].copy()
result_df.to_csv("data/agri_cluster_results.csv", index=False)
print("Saved: data/agri_cluster_results.csv")

# Agricultural Decision Recommendations
print("\n" + "="*65)
print("  AGRICULTURAL DECISION RECOMMENDATIONS PER K-MEANS ZONE")
print("="*65)
recs = {
    0: ("High-temp, high-rainfall rice belt (Terai).",
        "Focus: Blast & Sheath Rot prevention.",
        "Action: Improve drainage, reduce dense planting."),
    1: ("Cool, low-rainfall mixed-crop zone (Hilly/Mountain).",
        "Focus: Mosaic & Leaf Curl (viral diseases).",
        "Action: Aphid/whitefly control, use certified seeds."),
    2: ("Moderate mixed-crop flatland (Cotton, Mustard, Millet).",
        "Focus: Downy Mildew & Leaf Spot (fungal diseases).",
        "Action: Fungicide rotation, balanced NPK fertilizer."),
}
for zone in range(BEST_K):
    lines   = recs.get(zone, ("No specific recommendation.",))
    dom_reg = df[df['KMeans_Zone'] == zone]['region'].mode()[0]
    print(f"\n  Zone {zone}  ({dom_reg}):")
    for line in lines:
        print(f"    {line}")

print("\n✓ Farming region clustering complete.")
```
